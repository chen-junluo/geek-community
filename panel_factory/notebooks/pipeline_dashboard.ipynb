```python
import pandas as pd

def check_artifact(label, path):
    if not os.path.exists(path):
        print(f"[MISSING] {label}: {path}")
        return None
    df = pd.read_csv(path)
    n_rows, n_cols = df.shape
    print(f"[OK] {label}: {n_rows} rows × {n_cols} cols  →  {path}")
    return df

print("=== Features ===")
feature_artifacts = {
    "answer_llm_extension": check_artifact("answer_llm_extension", ARTIFACT_PATHS["features"]["answer_llm_extension"]),
    "answer_llm_deviation": check_artifact("answer_llm_deviation", ARTIFACT_PATHS["features"]["answer_llm_deviation"]),
    "question_ai_human_similarity": check_artifact("question_ai_human_similarity", ARTIFACT_PATHS["features"]["question_ai_human_similarity"]),
    "question_human_pairwise_similarity": check_artifact("question_human_pairwise_similarity", ARTIFACT_PATHS["features"]["question_human_pairwise_similarity"]),
}

print("\n=== Panels ===")
panel_artifacts = {
    "answer_panel": check_artifact("answer_panel", ARTIFACT_PATHS["panels"]["answer"]),
    "question_panel": check_artifact("question_panel", ARTIFACT_PATHS["panels"]["question"]),
}
```

```python
# answer_panel / question_panel 快速预览
print("=== answer_panel head ===")
display(answer_panel.head(3))

print("=== question_panel head ===")
display(question_panel.head(3))
```

```python
# Panel columns / row sanity checks
required_panel_columns = {
    "answer_panel": ["extension_score", "deviation_score"],
    "question_panel": [
        "human1_human2_similarity",
        "ai_human1_similarity",
        "ai_human2_similarity",
        "human_pairwise_similarity_mean",
    ],
}

for label, required_cols in required_panel_columns.items():
    df = panel_artifacts.get(label)
    if df is None:
        print(f"[SKIP] {label}: artifact missing")
        continue
    missing_cols = [c for c in required_cols if c not in df.columns]
    if missing_cols:
        print(f"[MISSING COLS] {label}: {missing_cols}")
    else:
        print(f"[OK] {label}: required columns present")

answer_intermediate = pd.read_csv(ARTIFACT_PATHS["intermediate"]["answer"])
question_intermediate = pd.read_csv(ARTIFACT_PATHS["intermediate"]["question"])
print(f"answer rows: intermediate={len(answer_intermediate)}, panel={len(answer_panel)}")
print(f"question rows: intermediate={len(question_intermediate)}, panel={len(question_panel)}")

# LLM feature 错误率检查
for label, df, score_col in [
    ("extension", feat_extension, "extension_score"),
    ("deviation", feat_deviation, "deviation_score"),
]:
    total = len(df)
    errors = df["error_reason"].notna().sum() if "error_reason" in df.columns else 0
    missing_score = df[score_col].isna().sum() if score_col in df.columns else "N/A"
    print(f"{label}: total={total}, errors={errors}, missing_score={missing_score}")
```
