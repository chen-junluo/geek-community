```python
from openai import OpenAI
import json
import os
from pprint import pprint
```

```python
API_KEY = "sk-27d25660b84405265f3b31f99bfb51c8fe8cd05a9f11392050e379085d648809"
API_BASE_URL = "https://claude.tokencode.fun/v1"
MODEL = "anthropic/claude-opus-4.7"

client = OpenAI(
    api_key=API_KEY,
    base_url=API_BASE_URL.rstrip("/"),
)

print({
    "base_url": API_BASE_URL,
    "model": MODEL,
})
```

```python
messages = [
    {
        "role": "system",
        "content": "You are Claude Opus 4.7. Answer briefly and clearly.",
    },
    {
        "role": "user",
        "content": "Reply with exactly: Opus 4.7 reached successfully.",
    },
]

response = client.chat.completions.create(
    model=MODEL,
    messages=messages,
    temperature=0,
    max_tokens=64,
)

response
```

```python
choice = response.choices[0]
content = choice.message.content

print("Model returned by provider:", getattr(response, "model", None))
print("\nAssistant content:")
print(content)
```

```python
# Optional: inspect the raw response payload if needed
try:
    pprint(response.model_dump())
except Exception as e:
    print("Could not dump full response:", e)
```

```python
# Optional fallback experiments if the first model id fails
candidate_models = [
    "anthropic/claude-opus-4.7",
    "claude-opus-4-7",
    "claude-opus-4.7",
    "opus-4.7",
]

results = []
for model_name in candidate_models:
    try:
        r = client.chat.completions.create(
            model=model_name,
            messages=[{"role": "user", "content": "Say OK."}],
            temperature=0,
            max_tokens=16,
        )
        results.append({
            "model_name": model_name,
            "ok": True,
            "returned_model": getattr(r, "model", None),
            "text": r.choices[0].message.content,
        })
    except Exception as e:
        results.append({
            "model_name": model_name,
            "ok": False,
            "error": str(e),
        })

results
```
